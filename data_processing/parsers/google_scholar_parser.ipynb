```python
df = pd.DataFrame(columns=['author', 'publication', 'data'])
```

```python
pubs_info = []
```

```python
pg = ProxyGenerator()
scholarly.use_proxy(pg, pg)
```

```python
for author_id in tqdm(author_ids):
    search_query = scholarly.search_author_id(author_id)
    author_info = scholarly.fill(search_query, sections=['publications'])
            
    for pub in tqdm(author_info['publications']):
        pub_filled = scholarly.fill(pub)
        try:
            author = pub_filled['bib'].get('author', 'Не указано')
        except Exception as e:
            author = f"Ошибка: {str(e)}"    
        try:
            title = pub_filled['bib'].get('title', 'Не указано')
        except Exception as e:
            title = f"Ошибка: {str(e)}"   
        try:
            data = pub_filled['bib'].get('pub_year', 'Не указано')
        except Exception as e:
            data = f"Ошибка: {str(e)}"

        pubs_info.append({
            'author': author,
            'publication': title,
            'data': data
        })

        time.sleep(random.uniform(1.1, 5.2))
        
    time.sleep(random.uniform(1.1, 5.2))
```

```python
pubs_info_df = pd.DataFrame(pubs_info)
```

```python
pubs_info_df.to_csv('publications_scholar.csv', index=False)
```
