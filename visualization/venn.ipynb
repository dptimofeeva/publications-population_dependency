```python
import pandas as pd
import matplotlib.pyplot as plt
from supervenn import supervenn
```

```python
df = pd.read_csv("publications_metadata.csv", sep=';')
```

```python
sets = [set(df[df["from_TALD"] == 1].index),
        set(df[df["from_Glottolog"] == 1].index),
        set(df[df["from_rsl"] == 1].index),
        set(df[df["from_Scholar"] == 1].index)
]
```

```python
labels = ['TALD', 'Glottolog', 'РГБ', 'Google Scholar']
colors = ['tab:green','tab:orange','tab:red','tab:blue']
```

```python
plt.figure(figsize=(20, 10))
supervenn(sets, 
          labels,
          widths_minmax_ratio=0.05,
          chunks_ordering='size',
          sets_ordering='size',
          reverse_sets_order=False,
          color_cycle=colors)
plt.xlabel('Количество публикаций', fontsize=16)
plt.ylabel('Источники', fontsize=16)
plt.show()
```
