```python
for i, row in df.iterrows():
    title_lower = str(row['title']).lower()
    languages_found = []
    
    for lang_name, variants in langs.items():
        for v in variants:
            if v in title_lower:
                languages_found.append(lang_name)
                break
    
    if languages_found:
        df.at[i, 'lang_TALD'] = ', '.join(languages_found)
    else:
        df.at[i, 'lang_TALD'] = 'Unknown'
```

```python
df.to_csv('publications_metadata.csv', index=False)
```
